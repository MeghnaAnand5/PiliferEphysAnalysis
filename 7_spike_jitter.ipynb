```python
# imports/mount Drive
import numpy as np
import matplotlib.pyplot as plt
import plotly.graph_objects as go
```

```python
# load data

# change path below to point to csv file location in Google Drive
file_folder = './dataFolder/5a_Raster_PSTH/'

# pathToData = file_folder + 'm31_unit_2_raster.csv'
# pathToData = file_folder + 'm11_unit_3_raster.csv'
pathToData = file_folder + 'm23_unit_1_raster.csv'


data = np.loadtxt(pathToData, delimiter=",", skiprows=1)
spikeTimes = data[:,1]
trialNum = data[:,0]
```

```python
# parameters for analysis

windowSize = 0.005  # in sec
windowStep = 0.001  # in sec, how much to step windows forward between possible events
eventThreshold = 0.8   # spikes must be present in this fraction of trials to call it an "event"

maxT = 10   # maximum time in recording, in sec
nTrials = 30  # maybe shouldn't hard-code this?
```

```python
AllSikesInWindow = []
AllSelectedSpikes = []
nWindows = int(np.floor(maxT/windowStep)-np.ceil(windowSize/windowStep)+1)
startT = 0
while startT <= (maxT-windowSize):
    spikesThisWindowIdx = (spikeTimes>=startT) & (spikeTimes<(startT+windowSize))
    
    if sum(spikesThisWindowIdx)>0:  # if any spikes in this possible window
        uniqueTrials, firstIdx = np.unique(trialNum[spikesThisWindowIdx], return_index=True)  # find all unique trials where there was a spike in this window
        if len(uniqueTrials) >= (nTrials*eventThreshold):
            selectedSpikes = spikeTimes[spikesThisWindowIdx][firstIdx]  # corrected to use only first spike on each trial that falls in this window
            meanTime = np.mean(selectedSpikes)
            jitterAbs = np.mean(np.abs(selectedSpikes-meanTime))
            jitterStd = np.std(selectedSpikes)
            
            print('mean time: ' + str(meanTime) + ' sec')
            print('jitterAbs: ' + str(np.round(jitterAbs*1000*1000)/1000) + ' msec')
            print('jitterStd: ' + str(np.round(jitterStd*1000*1000)/1000) + ' msec')
            print('frac trials: ' + str(len(uniqueTrials)/nTrials) )
            print('')

            startT += windowSize  # step forward a full window
            AllSikesInWindow.append(spikesThisWindowIdx)
            AllSelectedSpikes.append(selectedSpikes)
        else:
            startT += windowStep  # step forward with normal step size
    else:
        startT += windowSize  # step forward a full window
```

```python
# make zoomable raster plot

fig = go.Figure()

fig.add_trace(
    go.Scattergl(
        x=spikeTimes,
        y=trialNum+1,
        mode="markers",
        marker=dict(
            symbol="line-ns",   # vertical line marker
            size=10,            # controls spike height
            line=dict(width=1.5),
        ),
        hovertemplate=(
            "time: %{x:.3f} s<br>"
            "trial: %{y}<extra></extra>"
        ),
    )
)

fig.update_layout(
    template="simple_white",
    width=900,
    height=700,
    xaxis=dict(
        title="time (s)",
        range=[0, 10],
        showgrid=True,
        zeroline=False,
    ),
    yaxis=dict(
        title="trial",
        range=[0.5, 30.5],
        dtick=5,
    ),
)

fig.show()
```
