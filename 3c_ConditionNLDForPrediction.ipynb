### clips nld for forward prediction

#### bins beyond +- 2.5 z score are removed
#### bins with fewer than 200 spikes are also removed

```python
import os
import pandas as pd
import pickle
import glob
import numpy as np
```

```python
filepath_model = glob.glob('./dataFolder/5_1DCompModel/UnscaledFiringRate_NLD/v2/*.pickle')

out_path = './dataFolder/5_1DCompModel/Unscaled_clippedNLD/forForwardPrediction/v2/'
os.makedirs(out_path, exist_ok=True)                              
```

```python
for filename in filepath_model:
    name = os.path.basename(filename).split('_')[0]
    unitID = "_".join(os.path.basename(filename).split('_')[1:3])
    
    print(f'processing {name} and {unitID}')
    
    ########## get the data files #########################  
    with open(filename, 'rb') as f:
                savedModel = pickle.load(f)
    
    ste = savedModel[0]
    sta = savedModel[1]
    sta_error = savedModel[2]
    prior_ensemble = savedModel[3]
    prior_mean = savedModel[4]
    prior_error = savedModel[5]
    prior_std = savedModel[6]
    mean_fire_rate = savedModel[7]
    nld = savedModel[8]
    n_ste = savedModel[9]
    n_prior = savedModel[10]
    bin_centers = savedModel[11]
    
    
    true_arr = n_ste < 50
    max_index = np.max(np.where(~true_arr)) + 1
    clipped_bin = bin_centers[:-1][:max_index]
    clipped_nld = nld[:max_index]

    index = np.where((clipped_bin > -2.5) & (clipped_bin < 2.5))
    bin_fwdModel = clipped_bin[index]
    nld_fwdModel = clipped_nld[index]
        
     # save all data values
    saveObject = (ste, sta, sta_error, 
                  prior_ensemble, prior_mean, prior_error,
                  prior_std, mean_fire_rate,
                  nld, n_ste, n_prior, bin_centers,
                 bin_fwdModel, nld_fwdModel)
    with open(out_path + name + '_' + unitID + "_EncodingModel.pickle","wb") as f:
        pickle.dump(saveObject, f)
```
