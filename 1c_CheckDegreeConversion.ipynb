```python
import pandas as pd
import numpy as np
import glob
import matplotlib.pyplot as plt
import os
import pickle
```

```python
def makefig(moth, raw_volt, volt_scaled= None, degree=None):
    
    fig, axes = plt.subplots(3, 1, figsize=(10, 8), sharex=True)
    Fs = 3e4

    axes[0].plot(volt_raw, color='blue', linewidth=0.8)
    axes[0].set_ylabel('Raw Voltage')
    axes[0].set_title(f'Displacement for {moth} (Raw Voltage, 0-3.3V)')
    
    if volt_scaled:
        axes[1].plot(volt_scaled, color='blue', linewidth=0.8)
    axes[1].set_ylabel('Scaled Voltage')
    axes[1].set_title(f'Displacement for {moth} (Scaled Voltage, +-5V')
    
    if degree:
        axes[2].plot(disp_in_degree, color='yellow', linewidth=0.8)
    axes[2].set_title(f'Displacement for {moth} (Degrees)')
    axes[2].set_ylabel('Displacement (°)')
    axes[2].set_xlabel('Frame')

    for ax in axes:
        ax.axhline(0, color='gray', linewidth=0.5, linestyle='--')
#         ax.set_xlim(0, int(11.5 * Fs))  

    plt.tight_layout()
    plt.savefig(out_fig + moth + "_StimulusScaling.png")
    plt.close()
```

```python
# take raw voltage data, scaled and degree and plot them all together
```

```python
all_files = glob.glob('./dataFolder/4_RawData_Index_Timestamp/*.pickle')
convert2Deg_files = glob.glob('./dataFolder/4_RawData_Index_Timestamp/ScaledStimulus/*degrees.csv')

scaled = glob.glob('./dataFolder/4_RawData_Index_Timestamp/ScaledStimulus/*.pickle')

out_fig = "./dataFolder/Figures/1_Scaled_Stimulus/"
os.makedirs(out_fig, exist_ok=True)
```

```python
for file in all_files:
    print(f"processing {file}")
    moth = '_'.join(os.path.basename(file).split('_')[0:2])
    
    all_data_path = [f for f in convert2Deg_files if moth in f]
    
    if not all_data_path:
        # get raw stimulus
        rawfilepath = [file for file in all_files if moth in file][0]
        with open(rawfilepath, "rb") as f:
            rawdata = pickle.load(f)    
        volt_raw = rawdata[2].to_list()
        
        # get scaled stimulus
        rawfilepath = [file for file in scaled if moth in file][0]
        with open(rawfilepath, "rb") as f:
            rawdata = pickle.load(f)    
        volt_scaled = rawdata[2].to_list()
        makefig(moth, volt_raw, volt_scaled)
            
    else:
        all_data = pd.read_csv(all_data_path[0])
        volt_raw = all_data['raw_voltage'].to_list()
        volt_scaled = all_data['scaled_voltage'].to_list()
        disp_in_degree = all_data['deflectionAngle_degrees'].to_list()
        makefig(moth, volt_raw, volt_scaled, disp_in_degree)
```
