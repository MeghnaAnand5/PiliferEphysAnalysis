```python
import glob, pickle, os
import numpy as np

model_dir = './dataFolder/5_1DCompModel/UnscaledFiringRate_NLD/v2/'

for fp in sorted(glob.glob(model_dir + '*_EncodingModel.pickle')):
    with open(fp, 'rb') as f:
        sm = pickle.load(f)
    sta       = np.asarray(sm[1]).ravel()      # 660 samples
    prior_std = np.asarray(sm[5]).ravel()      # std per lag

    pre  = sta[:600]                            # before t=0
    peak = pre[np.argmax(np.abs(pre))]

    print(f'{os.path.basename(fp).replace("_EncodingModel.pickle",""):14s} '
          f'sta_peak={peak:+8.4f}   |sta_peak|={abs(peak):7.4f}   '
          f'max_prior_std={prior_std.max():7.4f}   ratio={abs(peak)/prior_std.max():.4f}')
```
