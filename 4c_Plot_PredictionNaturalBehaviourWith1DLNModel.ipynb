```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl
from scipy import interpolate
from scipy.stats import gaussian_kde
import glob
import pickle
import matplotlib
import os

matplotlib.rcParams['font.size'] = 8.0
matplotlib.rcParams['pdf.fonttype'] = 42
mpl.rcParams['font.family'] = 'Arial'
```

```python
Fs          = 3e4    #data sampling rate from orig
FS_ORIG     = 100    # natural behaviour video sampling rate
TimeHistory = 0.02   # 20ms pre-spike window (have not added post at all)
L           = int(TimeHistory * Fs)  # 600 samples, only consider length before 0

out_fig = "./dataFolder/Figures/4_NaturalBehavior/"
os.makedirs(out_fig, exist_ok=True)
```

```python
# 1D model files 
data_files = glob.glob("./dataFolder/6_NaturalBehavior/*_Prediction.pickle")
```

```python
data_files[0]
```

```python
for filepath in data_files:
    with open(filepath, 'rb') as file:
        savedData = pickle.load(file)
        
    nld_func = savedData[0]
    bin_centers = savedData[1]
    nld = savedData[2]
    proj = savedData[3]
    p_spike = savedData[4]
    spikes = savedData[5]
    upSampled_displacement = savedData[6]

    moth = os.path.basename(filepath).split('_')[0]
    unitID = os.path.basename(filepath).split('_')[1]
    beh_video = os.path.basename(filepath).split('_')[3]
    
    print(f'plotting data for {moth} {unitID} {beh_video}')
    
    f_nld = plt.figure()
    plt.plot(bin_centers[:-1], nld, 'o-')
    plt.plot(bin_centers[:-1], nld_func(bin_centers[:-1]))
    plt.savefig(out_fig + moth + '_' + unitID + '_'  + 'nld_func.png')
    plt.cla()
    
    plt.plot(np.arange(-10, 10), nld_func(np.arange(-10, 10)))
    plt.plot(np.arange(-4, 4), nld_func(np.arange(-4, 4)))
    plt.savefig(out_fig + moth + '_' + unitID + '_for_'  + 'nld_func_extended.png')
    f_nld.clf()
    
    f_pred, ax_pred = plt.subplots(4)
    ax_pred[0].plot(upSampled_displacement)
    ax_pred[0].set_title('Displacement (in degrees)')
    ax_pred[1].plot(proj)
    ax_pred[1].set_title('Projection (unitless)')  
    ax_pred[2].plot(p_spike)
    ax_pred[2].set_title('Prob (spike) (Hz)')
    
    indices = np.where(spikes)[0]
    # Plot vertical lines (ticks) at True indices
    # vlines(x, ymin, ymax)
    ax_pred[3].eventplot(indices)
#     , ymin=0, ymax=1, color='black', linestyle='-', label='True Ticks')
    f_pred.savefig(out_fig + moth + '_' + unitID + '_for_' + beh_video + 'prediction.png')
    f_pred.clf()
```
