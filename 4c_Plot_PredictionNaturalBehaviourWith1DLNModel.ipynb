```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl
from scipy import interpolate
from scipy.stats import gaussian_kde
import glob
import pickle
import matplotlib
import os

matplotlib.rcParams['font.size'] = 8.0
matplotlib.rcParams['pdf.fonttype'] = 42
mpl.rcParams['font.family'] = 'Arial'
```

```python
Fs          = 3e4    #data sampling rate from orig
FS_ORIG     = 100    # natural behaviour video sampling rate
TimeHistory = 0.02   # 20ms pre-spike window (have not added post at all)
L           = int(TimeHistory * Fs)  # 600 samples, only consider length before 0

out_fig = "./dataFolder/Figures/4_NaturalBehavior/clipped/v3/"
os.makedirs(out_fig, exist_ok=True)
```

```python
# 1D model files 
data_files = glob.glob("./dataFolder/6_NaturalBehavior/Clipped_test/v2/*_Prediction.pickle")
```

```python
SPIKE_MODE = 'max'  #change here to plot between max and mean

for filepath in data_files:
    with open(filepath, 'rb') as file:
        savedData = pickle.load(file)
        
    bin_fwdModel        = savedData[0]
    nld_fwdModel        = savedData[1]
    df_proj             = savedData[2]
    upSampled_displacement = savedData[3]

    moth = os.path.basename(filepath).split('_')[0]
    unitID = os.path.basename(filepath).split('_')[1]
    beh_video = os.path.basename(filepath).split('_')[3]
    
    print(f'plotting data for {moth} {unitID} {beh_video}')
    
    f_nld = plt.figure()
    plt.plot(bin_fwdModel, nld_fwdModel, 'o-')
    plt.savefig(out_fig + moth + '_' + unitID + '_'  + 'nld_func.png')
    plt.cla()
    
#     plt.plot(np.arange(-10, 10), nld_func(np.arange(-10, 10)))
#     plt.plot(np.arange(-4, 4), nld_func(np.arange(-4, 4)))
#     plt.savefig(out_fig + moth + '_' + unitID + '_for_'  + 'nld_func_extended.png')
#     f_nld.clf()
    
    f_pred, ax_pred = plt.subplots(4, 1, figsize=(12, 10), sharex=True)

    ax_pred[0].plot(upSampled_displacement, lw=0.6, color='black')
    ax_pred[0].set_title('Displacement (degrees)')
    ax_pred[0].set_ylabel('Degrees')

    ax_pred[1].plot(df_proj['proj'], lw=0.6, color='blue')
    ax_pred[1].set_title('Projection (z-score units)')
    ax_pred[1].set_ylabel('Projection')

    ax_pred[2].plot(df_proj[f'pred_{SPIKE_MODE}'], lw=0.6, color='orange')
    ax_pred[2].set_title(f'Predicted firing rate ({SPIKE_MODE})')
    ax_pred[2].set_ylabel('Rate')

    spike_times = np.where(df_proj[f'spikes_{SPIKE_MODE}'].values == 1)[0] / Fs
    ax_pred[3].vlines(spike_times, 0, 1, color='black', linewidth=0.5)
    ax_pred[3].set_title(f'Predicted spikes raster ({SPIKE_MODE})')
    ax_pred[3].set_ylabel('Spikes')
    ax_pred[3].set_yticks([])
    ax_pred[3].set_xlabel('Sample')

    plt.tight_layout()
    f_pred.savefig(out_fig + moth + '_' + unitID + '_for_' + beh_video + '_prediction.png')
    plt.close(f_pred)
```
